# Automotive Visualizations - Execution and Rendering Notebook

**Purpose:** Execute the automotive Python visualizations notebook and render all visualizations  
**Source Notebook:** `automotive_python_visualizations.ipynb`  
**Output:** Rendered visualizations with proper display and export capabilities

This notebook demonstrates how to programmatically execute another notebook and capture its visualization outputs for display and analysis.

```python
# Import Required Libraries for Notebook Execution
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from IPython.display import display, HTML, Image
import os
import warnings
warnings.filterwarnings('ignore')

print("✅ Notebook execution libraries imported successfully!")
print("📓 Ready to execute visualization notebook and capture outputs")
```

```python
# Set Up Notebook Execution Environment

# Define paths
notebook_path = r"d:\Seventh Semester\FOS\automotive-emissions-optimization\notebooks\automotive_python_visualizations.ipynb"
output_dir = r"d:\Seventh Semester\FOS\automotive-emissions-optimization\notebooks"

# Configure execution parameters
execution_timeout = 300  # 5 minutes timeout for each cell
kernel_name = 'python3'

print(f"📂 Source notebook: {os.path.basename(notebook_path)}")
print(f"📁 Output directory: {output_dir}")
print(f"⏱️ Execution timeout: {execution_timeout} seconds")
print(f"🐍 Kernel: {kernel_name}")

# Verify notebook exists
if os.path.exists(notebook_path):
    print("✅ Source notebook found and ready for execution")
else:
    print("❌ Source notebook not found - creating execution plan")
    print("   The visualization notebook will be executed when available")
```

```python
# Load and Execute External Notebook

def execute_visualization_notebook():
    """Execute the automotive visualizations notebook programmatically"""
    
    try:
        # Read the notebook
        with open(notebook_path, 'r', encoding='utf-8') as f:
            nb = nbformat.read(f, as_version=4)
        
        print(f"📖 Loaded notebook with {len(nb.cells)} cells")
        
        # Configure the executor
        ep = ExecutePreprocessor(timeout=execution_timeout, kernel_name=kernel_name)
        
        print("🚀 Starting notebook execution...")
        print("   This may take a few minutes to complete all visualizations")
        
        # Execute the notebook
        ep.preprocess(nb, {'metadata': {'path': output_dir}})
        
        print("✅ Notebook execution completed successfully!")
        
        # Count executed cells
        executed_cells = sum(1 for cell in nb.cells if cell.cell_type == 'code' and cell.get('outputs'))
        print(f"📊 Executed {executed_cells} code cells with outputs")
        
        return nb
        
    except FileNotFoundError:
        print("❌ Visualization notebook not found")
        print("   Please ensure automotive_python_visualizations.ipynb exists")
        return None
    except Exception as e:
        print(f"❌ Error during execution: {str(e)}")
        return None

# Execute the notebook
executed_notebook = execute_visualization_notebook()
```

```python
# Extract and Display Visualizations

def extract_visualization_outputs(notebook):
    """Extract and display visualization outputs from executed notebook"""
    
    if notebook is None:
        print("❌ No notebook to process")
        return
    
    visualization_count = 0
    
    print("🎨 EXTRACTING VISUALIZATION OUTPUTS")
    print("=" * 50)
    
    for i, cell in enumerate(notebook.cells):
        if cell.cell_type == 'code' and 'outputs' in cell:
            # Look for matplotlib figures or plotly outputs
            for output in cell.outputs:
                if output.output_type == 'display_data':
                    if 'image/png' in output.data:
                        visualization_count += 1
                        print(f"📊 Found visualization {visualization_count} in cell {i+1}")
                        
                        # Display the image
                        import base64
                        from IPython.display import Image, display
                        
                        img_data = base64.b64decode(output.data['image/png'])
                        display(Image(data=img_data))
                        
                    elif 'text/html' in output.data and 'plotly' in output.data['text/html']:
                        visualization_count += 1
                        print(f"📈 Found interactive visualization {visualization_count} in cell {i+1}")
                        
                        # Display the HTML
                        display(HTML(output.data['text/html']))
    
    print(f"\n✅ Successfully extracted and displayed {visualization_count} visualizations")
    return visualization_count

# Extract and display all visualizations
if executed_notebook:
    viz_count = extract_visualization_outputs(executed_notebook)
else:
    print("⚠️ Notebook execution required before extracting visualizations")
    viz_count = 0
```

```python
# Handle Multiple Plot Types and Save Generated Visualizations

def display_saved_visualizations():
    """Display saved visualization files from the executed notebook"""
    
    # List of expected visualization files
    viz_files = [
        'fuel_efficiency_dashboard.png',
        'quality_radar_chart.png', 
        'missing_value_treatment.png',
        'performance_analytics.png',
        'industry_benchmark_comparison.png'
    ]
    
    viz_titles = [
        'Figure 1: Fuel Efficiency Trends Dashboard',
        'Figure 2: Data Quality Assessment Radar Chart',
        'Figure 3: Missing Value Treatment Analysis', 
        'Figure 4: Processing Performance Analytics',
        'Figure 5: Industry Benchmark Comparison'
    ]
    
    print("🖼️ DISPLAYING SAVED VISUALIZATIONS")
    print("=" * 50)
    
    displayed_count = 0
    
    for i, (filename, title) in enumerate(zip(viz_files, viz_titles)):
        filepath = os.path.join(output_dir, filename)
        
        if os.path.exists(filepath):
            print(f"\n📊 {title}")
            print("-" * len(title))
            
            try:
                # Display the image
                img = mpimg.imread(filepath)
                plt.figure(figsize=(12, 8))
                plt.imshow(img)
                plt.axis('off')
                plt.title(title, fontsize=14, fontweight='bold', pad=20)
                plt.tight_layout()
                plt.show()
                
                displayed_count += 1
                print(f"✅ Successfully displayed: {filename}")
                
            except Exception as e:
                print(f"❌ Error displaying {filename}: {str(e)}")
        else:
            print(f"⚠️ File not found: {filename}")
            print(f"   Expected location: {filepath}")
    
    print(f"\n📈 VISUALIZATION SUMMARY")
    print("=" * 30)
    print(f"Total Expected: {len(viz_files)}")
    print(f"Successfully Displayed: {displayed_count}")
    print(f"Success Rate: {(displayed_count/len(viz_files)*100):.1f}%")
    
    if displayed_count == len(viz_files):
        print("🎉 ALL VISUALIZATIONS SUCCESSFULLY DISPLAYED!")
    else:
        print("⚠️ Some visualizations may need to be generated first")
        print("   Run the automotive_python_visualizations.ipynb notebook")
    
    return displayed_count

# Display all saved visualizations
displayed_viz_count = display_saved_visualizations()
```
